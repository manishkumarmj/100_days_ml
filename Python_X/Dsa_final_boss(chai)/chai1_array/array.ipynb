# memory level copy paste 

```python
import copy

# Nested list example
array1 = [[1, 2], [3, 4]]

# SHALLOW COPY ([:], list(), copy())
array2 = array1[:]
array1[0][0] = 99
print(array2)  # [[99, 2], [3, 4]]  ← changed! Because inner list is shared
print(" when i  make change in the array 1 then also changed in array 2 ")
print(array1)


# DEEP COPY
array1 = [[1, 2], [3, 4]]
array3 = copy.deepcopy(array1)
array1[0][0] = 99
print(array3)  # [[1, 2], [3, 4]]  ← unchanged! Complete independent copy
```

# Pop & Remove
