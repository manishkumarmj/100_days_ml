```python
class Point:
    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y
        print("Point created")
    def __str__ (self):
        return "({},{})".format(self.x_cod , self.y_cod)
```

```python
p = Point(3, 5)
print(p)  # Output: (3,5)
```

# How to access the attributes of a class?



```python
class Person:
    def __init__(self, name_input, contry_input):
        self.name = name_input
        self.contry = contry_input
    def greet(self):
        if (self.contry)== 'India':
            print("Nameste", self.name, "Welcome to india")
        else:
            print('Hello', self.name,'from',self.contry )


```

```python
p = Person('rahul', 'India')
print(p.name)
print(p.contry)
p.greet()
```

Key point: Python objects can get new attributes anytime - no need to define in __init__.

```python
p.gender="male"
```

```python
class Person:

    def __init__(self, name, gender):
        self.name = name 
        self.gender = gender
    def __str__(self):
        return f"{self.name} ({self.gender})"

p1= Person("ankit","male")
p2 = Person("rahul","sigma male ")
p3 = Person("mj", "Beta male")

l = [p1,p2,p3]
# Print each person using __str__
```

```python
# Print each person using __str__
for person in l:
    print(person)
```

```python
print(l)
for i in l:
    print(i.name,i.gender)
```
