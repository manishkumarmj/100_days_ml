```python
class node:
    def __init__(self,value=None ):
        self.data = value
        self.next = None
        self.prev = None
        print("node created")
    
class double_link():
    def __init__(self):
        self.head = None
    
    def insertAtEnd( self, value):
        temp = node(value) # create a temp varaibale whcih is pointed by temp 
        # and head pointer also pointing toward the 
        if (self.head == None ):
            self.head = temp # as our temp is none same gooes for the head
            return 
        # hat if there is already some element 
        # we can't move head 
        # we will use temp to move forward with temp untill we point towards t 
        t = self.head
        while (t.next != None):
            t = t.next
        t.next = temp
        temp.prev= t # &&&

    def InsertAtBegning(self,value):
        temp = node(value)
        # 1st --> check the head is none
        if (self.head == None):
            temp = node(value)  # create a temp varaibale whcih is pointed by temp
            return

        temp.next = self.head # t1 will be the head of our link list
        self.head.prev = temp
        self.head = temp


    def InsertAtmiddel(self, value, x):
        t = self.head
        # searching code 
        while(t.next != None):
            if(t.data==x):
                break
            else:
                t = t.next
        temp = node(value)
        # jab bhie change kare tho 1st temp me kare ge
        temp.next = t.next
        t.next.prev = temp.next
        t.next = temp
        temp.prev = t
    

    def delete(self,value):
        if( self .head == None):
            print( "linked list is empty")
            return
       # searching
        t = self.head
        # delete 1st node 
        if(t.data == value):
            self.head = t.next # will make head goes to next location  
            self.head.prev = None  # new node pointint toward the new head
            return
        #middel 
        while(t.next != None):
            if(t.data == value):
                t.prev.next = t.next
                t.next.prev = t.prev
                return
            else:
                t= t.next
                
    
        # last location
        if(t.data ==value):
            t.prev.next = None
            
                
            


    def print(self ):
        t1= self.head
        while(t1.next != None):
            print(t1.data,end = "<->")
            t1 = t1.next
        print(t1.data)



```
