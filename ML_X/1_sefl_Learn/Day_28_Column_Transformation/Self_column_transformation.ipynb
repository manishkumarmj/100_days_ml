# fill missing data

```python
# adding simple imputer to fever col
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])

# also the test data
X_test_fever = si.fit_transform(X_test[['fever']])
                                 
X_train_fever.shape
```

```python
# Ordinalencoding -> cough 
# mild --> low 
# STrong --> high 
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']]) # 80 % trained data 

# also the test data
X_test_cough = oe.fit_transform(X_test[['cough']]) # 20 % test data

X_train_cough.shape
```

# Gender se 1 column

```python
# OneHotEncoding -> gender,city
ohe = OneHotEncoder(drop='first',sparse_output=False)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])

# also the test data
X_test_gender_city = ohe.fit_transform(X_test[['gender','city']])

X_train_gender_city.shape
```

```python
# Extracting Age
X_train_age = X_train.drop(columns=['gender','fever','cough','city']).values

# also the test data
X_test_age = X_test.drop(columns=['gender','fever','cough','city']).values

X_train_age.shape
```

```python
from sklearn.compose import ColumnTransformer
```

# Create a object 
    * trasnformation of the data
    * you have 2  work
    * either u drop the data or passthough ( leave as it is ) 

    * Step 2
        * transformer passed in tuple 
        * I will pass 3 tuple tnf1 ... so on 
    * tnf1 --> fill the missing data 
    * tnf2 --> categories=[['Mild','Strong']]),['cough'] ranking of cough  on cough column
    * tnf3 --> OneHotEncoder(sparse=False,drop='first'),['gender','city'])
    

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']), # imputer fill ethe 
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```
