```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score

from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('spotify_churn_dataset.csv')[
['age','listening_time','songs_played_per_day','skip_rate','ads_listened_per_week','offline_listening', 
'is_churned']]
```

# for now i am picking up numberical data only

```python
df.head()
```

 * You can merger 2 column with each other here i won't see that so leave it   
 * df['Tracktime']= df['ofline_listening']+df['ads_listened_per_week']
 * Drop the data df.drop(columns=['SibSp','Parch'],inplace=True)

```python
df.head()
```

# prepare your data for X and Y 

```python
X = df.drop(columns=['is_churned']) # is churnmed is only removed 
y = df['is_churned'] # # is churnmed is only added test 
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.4,random_state=42)
```

```python
X_train.head()
```

# Without binarization


```python

clf = DecisionTreeClassifier()

clf.fit(X_train,y_train)

y_pred = clf.predict(X_test)

accuracy_score(y_test,y_pred)
```

# cross valadation 


```python
np.mean(cross_val_score(DecisionTreeClassifier(),X,y,cv=10, scoring='accuracy'))
```

# applying biniraztion 

```python
# Applying Binarization

from sklearn.preprocessing import Binarizer
```

Create a Tranfromer

```python
trf = ColumnTransformer([
    ('bin',Binarizer(copy=False),['age','listening_time','songs_played_per_day','skip_rate','ads_listened_per_week','offline_listening'])
],remainder='passthrough')
```

```python
X_train_trans=trf.fit_transform(X_train)
X_test_trans=trf.fit_transform(X_test)
```

```python
pd.DataFrame(X_train_trans,columns=['age','listening_time','songs_played_per_day','skip_rate','ads_listened_per_week','offline_listening'])
```

```python
clf = DecisionTreeClassifier()
clf.fit(X_train_trans,y_train)
y_updated_Pred = clf.predict(X_test_trans)
accuracy_score(y_test,y_updated_Pred)
```

```python
X_trf = trf.fit_transform(X)
np.mean(cross_val_score(DecisionTreeClassifier(),X_trf,y,cv=10,scoring='accuracy'))
```
