# As our data is shorted then we use train split test

```python
from sklearn.model_selection import train_test_split 
```

```python
spoti_Ilocked = spoti.iloc[:, [1, 4, 8]]  # Features
target = spoti.iloc[:, -1]                   # Target

X_train, X_test, Y_train, Y_test = train_test_split(
    spoti_Ilocked,  # Just variable, no '='
    target,         # Just variable, no '='
    test_size=0.2
)
#Separates features (X) from target (Y) for machine learning.
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

# Ranking the data 

```python
# According to my data  from from 29,28,27
oe = OrdinalEncoder(
    categories=[['Other','Male','Female'],
                 ['Family','Student','Free','Premium' ],
                   [ 'Mobile', 'Web','Desktop'], ]
    )  # Fit the ordinal encoder on our training set X_train oe.fit(X_train)
```

# train ke upper fit karte

```python
oe.fit(X_train)
```

# Tranform dnon ke upper  karte hai 

```python
X_train = oe.transform(X_train)
X_test= oe.transform(X_test)
```
