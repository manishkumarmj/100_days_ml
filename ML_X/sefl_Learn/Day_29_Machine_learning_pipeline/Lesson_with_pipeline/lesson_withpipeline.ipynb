```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('A:\\AIML_DEV_FOLDER\ML_X\\sefl_Learn\\Day_29_Machine_learning_pipeline\\train.csv')
```

```python
df.head()
```

# step 1 drop data
#### drop the uncessary data

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

# train test split 

```python
# Step 1 -> train/test/split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived']),
                                                 df['Survived'],
                                                 test_size=0.2,
                                                random_state=42)
```

# transfromer 
    * 1st for age 
    * 2nd embarked station 

# reaosn for numbmer 
    * when its transformed it converted in
    * DataFrame --> numpy array 
    * it will not rembember data name 

```python
# imputation transformer
trf1 = ColumnTransformer([
    ('impute_age',SimpleImputer(),[2]), # [number of colum' age  is 2 ] 
    ('impute_embarked',SimpleImputer(strategy='most_frequent'),[6])
],remainder='passthrough') 
```

# 2nd part me one hot encoding  
here we passed ohe together with column transformer  
1 for sex   
6 embarked station   

```python
trf2 = ColumnTransformer([
    ('ohe_sex_embarked',OneHotEncoder(sparse_output=False,handle_unknown='ignore'),[1,6])
],remainder='passthrough')
```

# min max scaller 

    * 1 total we have 7 column     
    *  sex male female 2  
    *  embarked 3 station 3   
    * 7- 2 ( sex , embarked ) + 2 + 3 = 10 

```python
 
# Scaling
trf3 = ColumnTransformer([
    ('scale',MinMaxScaler(),slice(0,10))
])
```

# feature scalling
k for top 5 function 

```python
# Feature selection
trf4 = SelectKBest(score_func=chi2,k=8)
```

```python
# train the model
trf5 = DecisionTreeClassifier()
```

# Create Pipeline

```python
pipe = Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
    ('trf3',trf3),
    ('trf4',trf4),
    ('trf5',trf5)
])
```

only in make_pipeline you don't need to send pipe line

# Pipeline Vs make_pipeline

Pipeline requires naming of steps, make_pipeline does not.

(Same applies to ColumnTransformer vs make_column_transformer)

```python
pipe = make_pipeline(trf1,trf2,trf3,trf4,trf5)
```

```python
# train
pipe.fit(X_train,y_train)
```

```python
# train
pipe.fit(X_train,y_train)
```

```python
# Predict
y_pred = pipe.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
# cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
cross_val_score(pipe, X_train, y_train, cv=5, scoring='accuracy').mean()
```
