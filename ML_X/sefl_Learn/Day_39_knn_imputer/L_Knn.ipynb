```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import KNNImputer,SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','Fare','Survived']]
```

```python
df.head()
```

```python
df.isnull().mean() * 100
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
X_train.head()
```

# Knn key points
 * missing_values : int, float, str, np.nan or None, default=np.nan
 The placeholder for the missing values. All occurrences of missing_values will
 be imputed. For pandas'dataframes with nullable integer dtypes with missing
 values, missing_values should be set to np.nan, since pd.NA will be converted
 to np.nan. 
 * n_neighbors : int, deface
 * weights : {'uniform', 'distance') or callable, default: 'uniform'
 * metric :'nan _ euclidean') or callable, default: 'nan _ euclidean'
 * add_indicator : boot, default=False

# code step orginal 
It creates a new pandas DataFrame with the transformed values, keeping the original column structure.

X_train.columns = ['age', 'salary']  
X_test_trf = [[0.5, 1.2]]  

Result DataFrame:  
   age  salary

# Step 1 
 * Createa object named with knn
 * now pass the x data for test and traiend
 * *yaha pe hamko har ek nebhiour ke liye loop* 
 * weight = unirom distance 

```python
knn = KNNImputer(n_neighbors=3,weights='distance')

X_train_trf = knn.fit_transform(X_train)
X_test_trf = knn.transform(X_test)
```

```python
pd.DataFrame(X_test_trf,columns=X_train.columns)
```

# logistic regression ko call kiya 
 * naya wala data fit kiya
 * predecit kiya hai 
 * accuracy score done

```python
lr = LogisticRegression()

lr.fit(X_train_trf,y_train)

y_pred = lr.predict(X_test_trf)

accuracy_score(y_test,y_pred)
```

# Comparision with Simple Imputer --> mean

```python
# Comparision with Simple Imputer --> mean

si = SimpleImputer()

X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
lr = LogisticRegression()

lr.fit(X_train_trf2,y_train)

y_pred2 = lr.predict(X_test_trf2)

accuracy_score(y_test,y_pred2)
```
