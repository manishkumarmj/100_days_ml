### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

````python
df  = pd.read_csv('https://tinyurl.com/preview/deprecated/2fe6vz4u')
df
````

# Querry will filter out all the data

Difference:

query() uses string — cleaner for complex conditions

Mask uses boolean Series — faster, more Pythonic

````python
mask = df["Country"] =="Iran"
iran_series2= df[mask]
iran_series2
````

````python
mask = df["Country"] =="Chna"
china_series2= df[mask]
china_series2
````

````python
iran_series = df.query(' Country ==  "Iran" ')
iran_series
````

````python
iran_series2.groupby('Year')["PM2.5"].sum()
````

````python
iran_series = df[df['Country'] == "Iran"].groupby('Year')['PM2.5'].sum()
china_series = df[df["Country"] == "China"].groupby("Year")['PM2.5'].sum()
````

````python
plt.plot(iran_series.index, iran_series.values, label="Iran's PM2.5", linestyle='--')
plt.plot(china_series.index, china_series.values, label="China's PM2.5", linestyle='-.')
plt.xlabel("year")
plt.xlabel("PM 2.5")
plt.grid()
plt.xticks(df['Year'].value_counts().index)
plt.legend()
plt.title("PM2.5 over the year ")
plt.show()
````

### **`Problem-2:`** Draw a histogram of the  column "PM10" of which the y-axis represents the probability (see the documentation how to draw the probability).
