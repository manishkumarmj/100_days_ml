# Matplotlib Linestyles

| Style | Code | Description |
|-------|------|-------------|
| `-` | `'solid'` | Solid line (default) |
| `--` | `'dashed'` | Dashed line |
| `-.` | `'dashdot'` | Dash-dot line |
| `:` | `'dotted'` | Dotted line |
| `None` | `'None'` | No line |

**Shorthand versions:**
```python
plt.plot(x, y, linestyle='solid')   # or '-'
plt.plot(x, y, linestyle='dashed')  # or '--'
plt.plot(x, y, linestyle='dashdot') # or '-.'
plt.plot(x, y, linestyle='dotted')  # or ':'

# Matplotlib Markers Reference

| Marker | Description | Code |
|--------|-------------|------|
| . | Point | `'.'` |
| , | Pixel | `','` |
| o | Circle | `'o'` |
| v | Triangle down | `'v'` |
| ^ | Triangle up | `'^'` |
| < | Triangle left | `'<'` |
| > | Triangle right | `'>'` |
| s | Square | `'s'` |
| p | Pentagon | `'p'` |
| * | Star | `'*'` |
| + | Plus | `'+'` |
| x | X | `'x'` |
| D | Diamond | `'D'` |
| d | Thin diamond | `'d'` |
| \| | Vertical line | `'\|'` |
| _ | Horizontal line | `'_'` |
| 1 | Tripod down | `'1'` |
| 2 | Tripod up | `'2'` |
| 3 | Tripod left | `'3'` |
| 4 | Tripod right | `'4'` |
| None | No marker | `'None'` |

**Example:**
```python
plt.scatter(x, y, marker='*', color='red')
plt.plot(x, y, marker='o', linestyle='--')

# Matplotlib 3D Scatter Markers

Same markers work for 2D and 3D scatter plots.

| Marker | Code | Description |
|--------|------|-------------|
| `.` | `'.'` | Point |
| `o` | `'o'` | Circle |
| `v` | `'v'` | Triangle down |
| `^` | `'^'` | Triangle up |
| `<` | `'<'` | Triangle left |
| `>` | `'>'` | Triangle right |
| `s` | `'s'` | Square |
| `p` | `'p'` | Pentagon |
| `*` | `'*'` | Star |
| `+` | `'+'` | Plus |
| `x` | `'x'` | X |
| `D` | `'D'` | Diamond |
| `d` | `'d'` | Thin diamond |
| `\|` | `'\|'` | Vertical line |
| `_` | `'_'` | Horizontal line |

**Example:**
```python
ax.scatter3D(x, y, z, marker='^', s=50, color='red')

# Matplotlib Colormaps (cmap)

## Sequential (one color intensity)
| cmap | Description |
|------|-------------|
| `'viridis'` | Blue-green-yellow (default) |
| `'plasma'` | Purple-red-yellow |
| `'inferno'` | Black-red-yellow |
| `'magma'` | Black-purple-yellow |
| `'cividis'` | Blue-yellow (colorblind-friendly) |
| `'Blues'` | Light to dark blue |
| `'Greens'` | Light to dark green |
| `'Reds'` | Light to dark red |
| `'Oranges'` | Light to dark orange |
| `'Purples'` | Light to dark purple |
| `'Greys'` | Light to dark gray |
| `'YlOrBr'` | Yellow-orange-brown |
| `'YlOrRd'` | Yellow-orange-red |

## Diverging (two colors with midpoint)
| cmap | Description |
|------|-------------|
| `'coolwarm'` | Blue-white-red |
| `'RdBu'` | Red-white-blue |
| `'RdYlBu'` | Red-yellow-blue |
| `'BrBG'` | Brown-white-green |
| `'PiYG'` | Pink-white-green |
| `'PRGn'` | Purple-white-green |
| `'Spectral'` | Multi-color spectrum |

## Qualitative (distinct colors)
| cmap | Description |
|------|-------------|
| `'Set1'` | 9 distinct colors |
| `'Set2'` | 8 softer distinct colors |
| `'Set3'` | 12 distinct colors |
| `'Pastel1'` | Pastel colors |
| `'Pastel2'` | Softer pastels |
| `'tab10'` | 10 distinct colors (default scatter) |
| `'tab20'` | 20 distinct colors |

## Cyclic
| cmap | Description |
|------|-------------|
| `'twilight'` | Continuous circular |
| `'hsv'` | Full color wheel |

**Usage:**
```python
ax.plot_surface(X, Y, Z, cmap='viridis')
plt.scatter(x, y, c=z, cmap='coolwarm')
