```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv('startup_funding.csv')
df.sample(10)
```

```python
df['Investors Name'].isna().sum()
```

```python
df['Investors Name']=df['Investors Name'].fillna('Undisclosed')
```

```python
sorted(df['Startup Name'].unique().tolist())
```

```python
df.info()
```

```python
df.set_index("Sr No", inplace = True)
```

```python
df.rename(columns={
    "Date":"date",
    "Startup Name":"starup",
    "Industry Vertical":"Vertical",
    "SubVertical":"subvertical",
    "City  Location":"city",
    "Investors Name":"investors",
    "InvestmentnType":"Round",
    "Amount in USD": "amount",

},inplace=True) 
```

```python
df.head()
```

# Convert  String amount --> int :



```python
df["amount"]=df["amount"].fillna('0') # undisclosed amount filled 
df["amount"]=df["amount"].str.replace(",","")
df["amount"]=df["amount"].str.replace("undisclosed","0")
df["amount"]=df["amount"].str.replace("Undisclosed","0")
```

```python
df = df[df["amount"].str.isdigit()]
```

```python
df["amount"] =df["amount"].astype('float64')
df["amount"] 
```

```python
df.info()
```

```python
df.head(5)
```

```python
def to_inr (dollar):
    inr=dollar *82.5
    return inr/10000000
```

```python
df["amount"]=df["amount"].apply(to_inr)
df["amount"]
```

```python
df.info()
```

```python
df["date"].shape
```

```python
df['date']=pd.to_datetime(df['date'])
```

```python
df['date'].shape
```

```python
df.info()
```

```python
df.drop('Remarks', axis=1, inplace=True)
```

```python
df.info()
```

```python
df = df.dropna(subset= ['date','starup','Vertical','city','investors','Round','amount'])
df.shape
```

```python
df.info()
```

```python
df.to_csv("final_version_1.csv",index=False)
```
