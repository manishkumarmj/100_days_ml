Plots under this section

```python
tips = sns.load_dataset('tips')
tips
```

# figure vs axis



```python
# scatter plot -> axes level function

sns.scatterplot(data = tips, x="total_bill" ,y= "tip",hue='sex' ,size='size')
```

# figure plot

```python
# scatter using relplot -> size and hue
```

```python


sns.relplot(data =tips, x="total_bill" ,y="tip" ,size='size', kind="scatter",hue="sex",style="time")
# relplot -> figure level -> square shape
```
