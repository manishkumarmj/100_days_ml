### `Q-1:` Using Gapminder Data
1. Create a scatter plot of 'gdpPercap' against 'lifeExp' for the year 2007, with the size of the markers determined by 'pop' and the color determined by 'continent'.

```python
import plotly.express as px
df = px.data.gapminder()
df.head()
```

```python
# Your Code here

data_2007= df[df["year"]==2007] 
data_2007


sns.relplot(data =data_2007, x="gdpPercap" ,y="lifeExp" ,size='pop', kind="scatter",hue="continent",)
```

```python
df = sns.load_dataset('flights')
df.tail()
```

## why we are using pivot here

### round  & astype()

df_pivot = df.pivot_table(index='year', columns='month', values='passengers', aggfunc='mean').round(0).astype(int)
 

```python
df_pivot=df.pivot_table(index='year', columns='month',values='passengers',aggfunc='mean')
df_pivot
```

```python
plt.figure(figsize=(25, 20))

sns.heatmap(df_pivot, annot=True, linewidths=0.6, cmap="summer", fmt='.0f',annot_kws={'size': 14})
```

```python
sns.clustermap(df_pivot,cmap="summer")
```
