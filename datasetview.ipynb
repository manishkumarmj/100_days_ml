```python
def text_sentiment_classifier(text):
    # Predefined keyword lists
    positive_keywords = ['good', 'great', 'excellent', 'amazing', 
                        'wonderful', 'happy', 'joy', 'love', 
                        'success', 'beautiful']
    
    negative_keywords = ['bad', 'terrible', 'awful', 'horrible', 
                        'sad', 'unhappy', 'hate', 'failure', 
                        'poor', 'ugly']
    
    # Convert text to lowercase for case-insensitive matching
    text_lower = text.lower()
    
    # Initialize counters
    positive_count = 0
    negative_count = 0
    
    # Count positive keywords
    for keyword in positive_keywords:
        if keyword in text_lower:
            # Count occurrences of each keyword
            positive_count += text_lower.count(keyword)
    
    # Count negative keywords
    for keyword in negative_keywords:
        if keyword in text_lower:
            # Count occurrences of each keyword
            negative_count += text_lower.count(keyword)
    
    # Determine sentiment
    if positive_count > negative_count:
        return "positive"
    elif negative_count > positive_count:
        return "negative"
    else:
        return "neutral"

# Test with the provided examples
example1 = "Today was a great day! I feel happy and the weather was wonderful."
print(f"Input: {example1}")
print(f"Output: {text_sentiment_classifier(example1)}")  # Should output: positive

print("\n---\n")

example2 = "The movie was terrible and boring. The plot was bad, but the actors were good."
print(f"Input: {example2}")
print(f"Output: {text_sentiment_classifier(example2)}")  # Should output: negative

# Additional tests
test_cases = [
    ("I feel nothing special about today.", "neutral"),
    ("I love this amazing product! It's excellent and wonderful!", "positive"),
    ("This is bad and awful. I hate it so much.", "negative"),
    ("Good but also bad at times.", "neutral"),  # Equal counts
]

print("\n---\nAdditional Tests:")
for text, expected in test_cases:
    result = text_sentiment_classifier(text)
    status = "✓" if result == expected else "✗"
    print(f"{status} Input: {text}")
    print(f"  Expected: {expected}, Got: {result}")
```
