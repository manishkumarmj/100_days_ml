````python
ipl = ipl.replace('Delhi Daredevils', 'Delhi Capitals')
ipl = ipl.replace('Kings XI Punjab', 'Punjab Kings')
ipl = ipl.replace('Rising Pune Supergiants', 'Rising Pune Supergiant')
ipl.sample(10)
````

###`Q-6` Write a code which can display the bar chart of top 5 teams who have played maximum number of matches in the IPL.
>  Hint: Be careful the data is divided in 2 different cols(Team 1 and Team 2)


1st we  sum all the data 
 - pd series= me bana dege

#### 

````python
total_series = pd.Series(pd.concat([ipl['Team1'], ipl['Team2']], ignore_index=True), name='total team')
total_series.value_counts()
total_series.value_counts(ascending=False)
total_series.value_counts(ascending=False).head(5)
total_series.value_counts(ascending=False).head(5).plot(kind='bar')
````
