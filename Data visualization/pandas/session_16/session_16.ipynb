# custom index
- here you can set the index of your own

```python

marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

# setting a name

name='Nitish ke marks'

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='Nitish ke marks')
marks
```
