#### list --> array 

#### series =  index + list

we will make series using list

```python
# string
country = ['India','Pakistan','USA','Nepal','Srilanka','lundman']

pd.Series(country) # serize bole ke class hai usme object contry de rah hai 
# 
```

```python
# integers
runs = [13,24,56,78,100]

runs_ser = pd.Series(runs)
```

# custom index
- here you can set the index of your own

```python

marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

# setting a name

name='Nitish ke marks'

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='Nitish ke marks')
marks
```

### Series from dict

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}
```

```python

marks_series = pd.Series(marks,name='Anpadh')
marks_series
```
