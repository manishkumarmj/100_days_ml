###`Q-5` Can you create a identity mattrix of shape (3,4). If yes write code for it.

````python
#Code here
iden_matrix= np.ones((3,3)) 
iden_matrix[0, 1:3] = 0
iden_matrix[1, 0:3:2] = 0
iden_matrix[2, 0:2] = 0

iden_matrix
````

point = np.array([2,3])
