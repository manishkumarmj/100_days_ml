````python
# code here

a = np.array([[3, 2, np.nan, 1], [10, 12, 10, 9], [5, np.nan, 1, np.nan]])



b = np.isnan(a) 
print(b) 

a[np.isnan(a)] = 0 # vector from 
print(a)
````

````python

a = np.array([[3, 2, np.nan, 1], [10, 12, 10, 9], [5, np.nan, 1, np.nan]])

for i, row in enumerate(a):
    for j, val in enumerate(row):
        if np.isnan(val):  # Use np.isnan() not ==
            a[i][j] = 0

print(a)
````
