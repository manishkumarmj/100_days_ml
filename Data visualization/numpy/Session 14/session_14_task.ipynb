```
x = numpy.array([1,2,3,4]).reshape(-1, 1)
y = numpy.array([5,6,7])
```

`Cij =1/(xi - yj)`

````python
x = np.array([1,2,3,4]).reshape(-1, 1) # -1 will make arrray 2d 
y = np.array([5,6,7])

arr = np.empty((len(x),len(y)), dtype=object)


for row in range(len(arr)):
    for col in range(len(arr[row])):
        value =  int(x[row][0] - y[col])
        arr[row, col] = round((1/value),2) #round off the value of updato 2 digit 
        
arr

    
````

[row][0] does two indexing operations:

[row] - gets the row as a 1D array

[0] - gets the first element from that row

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$
**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
import math

x = np.array([5, 6, 7], dtype=float) 
for i in range(len(x)):
    
    num = math.exp(x[i]) - math.exp(-x[i])
    deno = math.exp(x[i]) + math.exp(-x[i])
    x[i] = (num/deno)

x 
````
