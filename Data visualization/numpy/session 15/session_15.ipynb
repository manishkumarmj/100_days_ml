### np.sort

Return a sorted copy of an array.

https://numpy.org/doc/stable/reference/generated/numpy.sort.html

```python
# code
import numpy as np
a = np.random.randint(1,100,15)
a
```

```python
b = np.random.randint(1,100,24).reshape(6,4)
b
```

#### sort
- sorted --> list return 
- np.sort --> numpy's array return


- numpy.sort( user_array, axis =1/0, kind= none , order = None)

- kind  --> give u diffrent sorting algo like, heap sort , merge sort , quick sort, etc 





```python
np.sort(a)[::-1]
```

```python
print(b)
print('\n')
print(np.sort(b,axis =0)) # 0 for column
```

```python
print(b)
print('\n')
print(np.sort(b,axis =1)) # row wise 
```

### adding random value to row 

## breakdown 


- 0        → low (minimum value, inclusive)
- 10       → high (maximum value, exclusive) → generates 0 to 9
-size     → shape of output array
- b.shape[0] → number of rows in b (e.g., 2)
- 4           → 4 columns

### No. shape[0] = rows, shape[1] = columns, shape[2] = depth (3D arrays).

## 2D array
- b.shape[0]  # rows
- b.shape[1]  # columns

## 3D array
- b.shape[0]  # depth
- b.shape[1]  # rows  
- b.shape[2]  # columns
