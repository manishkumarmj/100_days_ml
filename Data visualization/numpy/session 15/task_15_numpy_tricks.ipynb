### `Q-5:` Get row numbers of NumPy array having element larger than X.
```
arr = [[1,2,3,4,5],
      [10,-3,30,4,5],
      [3,2,5,-4,5],
      [9,7,3,6,5]]

X = 6
```

````python
arr = np.array( [[1,2,3,4,56],
                [10,-3,30,4,5],
                [3,2,5,-4,5],
                [9,7,3,6,5]] )
result=[]
x = 6
for row in range(len(arr)):
    for col in range(len(arr[row])):
        if x <arr[row][col]:
            result.append(row)
result = np.array(result)
result = np.unique(result)
result
````
