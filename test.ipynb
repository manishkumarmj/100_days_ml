```python
#code here
import numpy as np
points = np.random.randint(1, 100, size=(10,2))
given_point=np.array([[10,2]])
distance = np.zeros((1,10))

for i in range(len(points)):
    distance[0, i] = np.sqrt(((points[i,0] - given_point[0,0])**2 + 
                              (points[i,1] - given_point[0,1])**2))
print(distance)
```
