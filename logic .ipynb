# Shorting alphabetic

```python
# Sort strings alphabetically

# 1. sorted() - returns new list
words = ["banana", "apple", "cherry"]
sorted_words = sorted(words)  # ['apple', 'banana', 'cherry']

# 2. sort() - sorts in-place
words.sort()  # words becomes ['apple', 'banana', 'cherry']

# 3. Sort string characters
s = "hello"
sorted_chars = sorted(s)  # ['e', 'h', 'l', 'l', 'o']
result = ''.join(sorted_chars)  # 'ehllo'

# 4. Case-sensitive (uppercase comes first)
words = ["Apple", "banana", "Cherry"]
sorted(words)  # ['Apple', 'Cherry', 'banana']

# 5. Case-insensitive sort
sorted(words, key=str.lower)  # ['Apple', 'banana', 'Cherry']
```

# couning a number

```python
# For string
s = "123456"
count = len(s)  # 6

# For list
arr = [1,2,3,45,6]
count = len(arr)  # 5

# For integer (convert to string first)
num = 123456
count = len(str(num))  # 6

# Manual count (string)
s = "123456"
count = 0
for char in s:
    count += 1
# count = 6

# Manual count (list)
arr = [1,2,3,45,6]
count = 0
for item in arr:
    count += 1
# count = 5

# Manual count (integer)
num = 123456
count = 0
temp = num
while temp > 0:
    count += 1
    temp = temp // 10
# count = 6
```

# reverse the number


```python
nums = [1,2,3,4,5,6,7,8]

# Method 1: reversed()
for num in reversed(nums):
    print(num)

# Method 2: negative step
for i in range(len(nums)-1, -1, -1):
    print(nums[i])

# Method 3: slicing
for num in nums[::-1]:
    print(num)

# Method 4: reverse then loop
nums.reverse()
for num in nums:
    print(num)
```

# loops 

```python
test_list = [10, 20, 30, 40]

# Use when you need INDEX position:
for i in range(len(test_list)):
    print(i, test_list[i])  # i = 0,1,2,3

# Use when you need VALUES only:
for i in test_list:
    print(i)  # i = 10,20,30,40

# Use when you need BOTH index and value:
for i, val in enumerate(test_list):
    print(i, val)
```
