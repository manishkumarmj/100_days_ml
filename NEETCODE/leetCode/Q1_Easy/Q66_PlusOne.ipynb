#logic for beaking up addding digit 


```python
nums = [9,9,9]
nums[-1]= nums[-1]+1
for i in range(len(nums)-1, -1, -1):
    #print(nums[i])
    temp = nums[i]
    # this is for inserting for code from last digit 
    # step 1 --> we check the leght 
    # step 2 we check if its not for the length 0 
    # Step 3 --> we check the last digit using %
    # step 4 --> we check the left most digit 
    # then final step checking the last  0th index

    while len(str(temp)) >1 :
        if( nums[0]!= temp):
            nums[i]= temp%10  # get the last didigit
            temp = temp//10  # get the left most digit 
            nums[i-1] = temp +nums[i-1]
        else: 
            # get the last didigit 
            temp1 = temp%10  
            nums[0] = temp1
            temp2 = temp //10 
            nums.insert(nums[0],temp2)
        break
    break
            
print(nums)


```

```python
nums = [9,9,9]
nums[-1] = nums[-1] + 1

i = len(nums) - 1 
while i >= 0:
    if nums[i] >= 10:
        carry = nums[i] // 10
        #// divied 
        nums[i] = nums[i] % 10
        # gives reminder
        if i == 0:
            nums.insert(0, carry)
        else:
            nums[i-1] = nums[i-1] + carry
    i -= 1 #i = i - 1

print(nums)  # [1,0,0,0]
```
