```python
input="anagram"
output="nagaram"
```

# sorted() automatically converts string to list of characters

```python
char_sort=sorted(input)
print(char_sort)
char_sort = ''.join(char_sort)
print(char_sort)
```

```python
char_out=sorted(output)
print(char_out)
char_out = "".join(char_out)
print(char_out)

```

```python
if (char_sort==char_out):
    print("true")
else:
    print("False")

```
