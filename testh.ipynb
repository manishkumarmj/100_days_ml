```python
s = "elloWorldPython"
result = []
temp = ""

for char in s:
    if char.isupper() and temp:
        result.append(temp)
        temp = char
    else:
        temp += char
result.append(temp)

print(result)  # ['Hello', 'World', 'Python']
```
