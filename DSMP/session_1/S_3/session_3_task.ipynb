### `Problem 1` - Print the following pattern. Write a program to use for loop to print the following reverse number pattern.

```bash
5 4 3 2 1
4 3 2 1
3 2 1
2 1
1
```

````python
nums = 5
for row in range(nums,0,-1):
    for col in range( row,0,-1 ):
        print(col,end= " " )
        col = col-1
    print(end= "\n")
````

###`Problem 8`: Write a program to print all the unique combinations of 1,2,3 and 4

`Output`:
```
1 2 3 4
1 2 4 3
1 3 2 4
1 3 4 2
1 4 2 3
1 4 3 2
2 1 3 4
2 1 4 3
2 3 1 4
2 3 4 1
2 4 1 3
.
.
and so on
```

With 4 digits:
- Repetition allowed: 10⁴ = 10,000 combinations (0000 to 9999)
- No repetition: 10 × 9 × 8 × 7 = 5,040 combinations

P(n, r) = n! / (n-r)!

Where:
n = total items (10 digits: 0-9)
r = items to choose (4 digits)

P(10,4) = 10! / (10-4)!
        = 10! / 6!
        = 10 × 9 × 8 × 7
        = 5,040

n = 4 (digits 1, 2, 3, 4)
r = 4 (using all digits)

P(4,4) = 4! / (4-4)!
       = 24 / 0!
       = 24 / 1
       = 24

# Total combinations: 24 (4! = 24)

````python
from itertools import permutations
i =1
for combo in permutations([1, 2, 3, 4]):
    print(i,"-->" ,*combo )# With >> # 1 2 3 4
    i+= 1
````

###`Problem 9`: Write a program that will take a decimal number as input and prints out the binary equivalent of the number

````python
# Code here
nums = 5 #int(input("Enter the decimal number "))

s_reminder= " "

while nums >0:
    reminder = nums % 2
    s_reminder = str(s_reminder)+str(reminder)
    print(s_reminder)
    nums = nums // 2
    
s_reminder=s_reminder[::-1]
print(s_reminder)
````
